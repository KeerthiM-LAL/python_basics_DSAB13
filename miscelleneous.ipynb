```python
class Employee:
    def __init__(self):
      self.name = "Rahul"    #public
      self._salary = 50000   #protected
      self._password = "12345" #private

    def show_details(self):
      print(self.name)
      print(self._salary)
      print(self._password)

emp = Employee()
print(emp.name)       #works
print(emp._salary)    #works, but discouraged
#print(emp._password) #error
emp.show_details()
```

```python
class Employee:
    def __init__(self):
      self.name = "Rahul"    #public
      self._salary = 50000   #protected
      self.__password = "12345" #private
    def show_parent(self):
      print("Name:", self.name)
      print("Salary:", self._salary)
      print("Password:", self.__password)

class Manager (Employee):
  def show_manager(self):
    print("Name:", self.name)
    print("Salary:", self._salary)

manager = Manager()
manager.show_manager()
```
